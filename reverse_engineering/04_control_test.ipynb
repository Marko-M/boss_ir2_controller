# 🎸 BOSS IR2 - Control Test

Testiamo la libreria `boss_ir2_control.py` per inviare comandi al pedale.

```python
from boss_ir2_control import BossIR2
import time

# Inizializza connessione
pedal = BossIR2()
```

## 1. Test Cambio Amplificatore (MODEL)
Dovresti vedere il LED sul pedale cambiare colore/posizione.

```python
models = pedal.get_models_list()
print("Modelli disponibili:", models)

# Prova tutti i modelli in sequenza
for model in models:
    pedal.set_param("MODEL", model)
    time.sleep(1.5)
```

## 2. Test Equalizzazione
Muoviamo i controlli EQ.

```python
# Porta tutto a 0
print("EQ -> 0")
pedal.set_param("BASS", 0)
pedal.set_param("MIDDLE", 0)
pedal.set_param("TREBLE", 0)
time.sleep(1)

# Porta tutto a metà (64)
print("EQ -> 50% (64)")
pedal.set_param("BASS", 64)
pedal.set_param("MIDDLE", 64)
pedal.set_param("TREBLE", 64)
time.sleep(1)

# Porta tutto al massimo (127)
print("EQ -> 100% (127)")
pedal.set_param("BASS", 127)
pedal.set_param("MIDDLE", 127)
pedal.set_param("TREBLE", 127)
```
