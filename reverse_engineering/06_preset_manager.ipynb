# 🎸 BOSS IR-2 - Manager Completo

Usa questo notebook per gestire i tuoi preset!

Funzionalità:
1. **BACKUP**: Salva tutti i parametri attuali su file
2. **RESTORE**: Ricarica un preset salvato

```python
from boss_ir2_tool import BossIR2Manager
import glob
import json

# Inizializza
pedal = BossIR2Manager()
```

## 💾 Backup (Salva Preset)

```python
preset = pedal.dump_preset()

# Salva su file con nome personalizzato
NAME = "My_Clean_Tone"  # ⬅️ Dai un nome al preset!

filename = f"presets/{NAME}.json"
import os
os.makedirs("presets", exist_ok=True)

with open(filename, 'w') as f:
    json.dump(preset, f, indent=2)

print(f"\n✅ Preset salvato in: {filename}")
```

## 📂 Restore (Carica Preset)

```python
# Lista preset disponibili
print("Preset salvati:")
files = glob.glob("presets/*.json")
for i, f in enumerate(files):
    print(f"  {i}. {os.path.basename(f)}")

# Scegli quale caricare
# IDX = 0  # ⬅️ Metti l'indice qui
# pedal.load_preset(files[IDX])
```
